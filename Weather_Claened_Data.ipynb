```python
numeric_columns=df.select_dtypes(include=[np.number]).columns
categorical_columns=df.select_dtypes(exclude=[np.number]).columns

for col in numeric_columns:
  if df[col].isnull().sum()>0:
    mean_val=df[col].mean()
    df[col]=df[col].fillna(mean_val)
    print(f"Filled {col} nulls with mean = {mean_val:.2f}")

for col in categorical_columns:
  if df[col].isnull().sum()>0:
    mode_val=df[col].mode()[0]
    df[col]=df[col].fillna(mode_val)
print("\nRemaining nulls after filling: ",df.isnull().sum().sum())
```

```python
num_duplicates= df.duplicated().sum()
print("Duplicate rows found: ", num_duplicates)
```

```python
df.drop_duplicates(inplace=True)
```

```python
num_duplicates= df.duplicated().sum()
print("Duplicate rows found: ", num_duplicates)
```

```python
import pandas as pd

df['Date'] = pd.to_datetime(df['Date'])

int_cols = ['Year', 'Month', 'Day', 'Weather Code']

for col in int_cols:
    df[col] = df[col].round().astype('Int64')

cat_cols = ['City', 'Season']
for col in cat_cols:
    df[col] = df[col].astype('category')
```

```python
numeric_cols_to_check = [
    'Temperature Max', 'Temperature Min', 'Temperature Mean',
    'Apparent Temperature', 'Relative Humidity', 'Precipitation',
    'Wind Speed', 'Wind Direction', 'Surface Pressure',
    'Cloud Cover', 'Sunshine Duration'
]
outlier_bounds = {}
for col in numeric_cols_to_check:
    q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outlier_bounds[col] = (lower, upper)
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: bounds=({lower:.2f}, {upper:.2f}) -> {n_outliers} outlier(s)")
```

```python
before = len(df)
outlier_mask = pd.Series(False, index=df.index)
for col, (lower, upper) in outlier_bounds.items():
    outlier_mask |= (df[col] < lower) | (df[col] > upper)
df = df[~outlier_mask].reset_index(drop=True)
after = len(df)
print(f"Rows before: {before} | Rows after removing IQR outliers: {after} | Removed: {before - after}")
```
